# Data preprocessing  :

## Removing grade columns

Columns G1, G2, and G3 were removed.

Reason:
These grades directly reflect academic performance and strongly determine the target variable (`pass`).

Keeping them would introduce leakage and make prediction unrealistically easy.

```python
df = df.drop('G3', axis=1)
df = df.drop('G1', axis=1)
df = df.drop('G2', axis=1)
```

## Outlier handling

The `absences` variable showed extreme values.

Method:
- IQR detection
- clipping upper extremes
- log transformation

Reason:
This reduces skewness while preserving ordering, helping scaling-based models perform better.

```python
threshold = 0.05   # 5%

cat_cols = df.select_dtypes(include='object').columns

rare_categories = {}

for col in cat_cols:
    freq = df[col].value_counts(normalize=True)
    rare = freq[freq < threshold]

    if not rare.empty:
        rare_categories[col] = rare

for col, rare in rare_categories.items():
    print(f"\nColumn: {col}")
    print("might have outliers categories:")
    print(rare)
```

```python
Q1 = df['absences'].quantile(0.25)
Q3 = df['absences'].quantile(0.75)
IQR = Q3 - Q1

lower = 0
upper = Q3 + 1.5 * IQR
print(lower)
print(upper)
```

```python
outliers = df[(df['absences'] < lower) | (df['absences'] > upper)]
print(outliers)
```

```python
df['absences'] = np.clip(df['absences'], lower, upper)
```

## Train-test split

A stratified 80/20 split was used.

Reason:
Stratification preserves class proportions in both train and test sets, which is important for imbalanced classification.

```python
from sklearn.model_selection import train_test_split

X = df.drop(['pass'], axis=1)
y = df['pass']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
y_train.shape
```

## One-hot encoding

Categorical variables were encoded using one-hot encoding.

Reason:
Most categorical variables are nominal (no natural order), so one-hot encoding avoids introducing false ordinal relationships.

```python
# One-Hot for categorical (nominal)
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

# align columns (VERY IMPORTANT)
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)
```

OneHotEncoder was preferred over pd.get_dummies() because it learns category mappings only from the training set, then applies the same mapping to test data. This prevents leakage and safely handles unseen categories using handle_unknown='ignore'. It is a more robust and production-ready encoding method

## Class balancing using SMOTE

SMOTE was applied only on the training set.

Reason:
The dataset is imbalanced.

SMOTE creates synthetic minority samples, improving recall and allowing models to learn minority-class patterns better.

It was applied only on training data to avoid leakage.

```python
print(pd.Series(y_train).value_counts())
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
```

```python
print(y_train.value_counts())        # before
print(y_train_res.value_counts())    # after
```

## Feature scaling

StandardScaler was applied after balancing.

Reason:
Scaling is important for:
- Logistic Regression
- KNN

because both depend on feature magnitude.

```python
original_feature_names = X.columns.tolist()
print(original_feature_names)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_res_scaled = scaler.fit_transform(X_train_res)
X_test_scaled = scaler.transform(X_test)
```

## Feature selection

SelectKBest with ANOVA F-score was used.

Reason:
It keeps the most statistically informative features while removing noise.

Benefits:
- reduces dimensionality,
- improves generalization,
- reduces overfitting risk.

```python
from sklearn.feature_selection import SelectKBest, f_classif
selector = SelectKBest(f_classif, k=20)
X_train_selected = selector.fit_transform(X_train_res_scaled, y_train_res)
X_test_selected  = selector.transform(X_test_scaled)
```

```python
scores = selector.scores_

feature_scores = (
    pd.DataFrame({
        'Feature': X_train_res.columns,
        'Score': scores
    })
    .sort_values('Score', ascending=False)
)

print(feature_scores.head(20))
```

#Model training :
Four models were selected to compare different learning approaches:

- Logistic Regression → linear baseline
- Decision Tree → interpretable nonlinear model
- Random Forest → ensemble model with strong generalization
- KNN → distance-based classifier

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
from sklearn.feature_selection import SelectKBest, f_classif
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, C=0.1),
    "Decision Tree":       DecisionTreeClassifier(max_depth=4, min_samples_split=15, random_state=42),
    "Random Forest":       RandomForestClassifier(max_depth=6, n_estimators=100,
                                                   min_samples_split=10,
                                                   max_features='sqrt',
                                                   random_state=42),
    "KNN":                 KNeighborsClassifier(n_neighbors=9)
}

results = {}
```

# Model Evaluation

Metrics used:

Accuracy:
overall correctness

Precision:
how reliable positive predictions are

Recall:
ability to detect positives

F1-score:
balance between precision and recall

ROC-AUC:
ranking quality

Confusion Matrix:
error type analysis

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix

def evaluate(model, X_tr, y_tr, X_te, y_te):
    y_pred  = model.predict(X_te)
    y_proba = model.predict_proba(X_te)[:, 1]

    train_acc = accuracy_score(y_tr, model.predict(X_tr))
    test_acc  = accuracy_score(y_te, y_pred)
    cm = confusion_matrix(y_te, y_pred)

    return {
        "Train Acc":  round(train_acc, 4),
        "Test Acc":   round(test_acc, 4),
        "Precision":  round(precision_score(y_te, y_pred), 4),
        "Recall":     round(recall_score(y_te, y_pred), 4),
        "F1":         round(f1_score(y_te, y_pred), 4),
        "ROC-AUC":    round(roc_auc_score(y_te, y_proba), 4),
        "Confusion Matrix": cm
    }
```

```python
results = {}

for name, model in models.items():
    model.fit(X_train_selected, y_train_res)

    results[name] = evaluate(
        model,
        X_train_selected, y_train_res,
        X_test_selected, y_test
    )
```

```python
import pandas as pd

df_results = pd.DataFrame(results).T
print(df_results.drop(columns=["Confusion Matrix"]))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

for name in results:
    plt.figure()
    sns.heatmap(results[name]["Confusion Matrix"], annot=True, fmt='d')
    plt.title(name)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
```

## Cross-validation analysis

Random Forest achieved the highest mean score, indicating strongest overall generalization.

KNN showed the lowest variance, meaning it was the most stable across folds.

```python
from sklearn.model_selection import cross_val_score

print("\n── Cross-Validation (5-fold) ──")
for name, model in models.items():
    scores = cross_val_score(model, X_train_res_scaled, y_train_res, cv=5)
    print(f"{name:25s}  CV Mean: {scores.mean():.4f}  ±{scores.std():.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve

def plot_learning_curve(model, X, y, title="Learning Curve"):
    train_sizes, train_scores, val_scores = learning_curve(
        model,
        X,
        y,
        cv=5,
        scoring="accuracy",
        train_sizes=np.linspace(0.1, 1.0, 10),
        n_jobs=-1
    )

    train_mean = train_scores.mean(axis=1)
    val_mean = val_scores.mean(axis=1)

    plt.figure(figsize=(8,5))
    plt.plot(train_sizes, train_mean, label="Training Accuracy")
    plt.plot(train_sizes, val_mean, label="Validation Accuracy")

    plt.title(title)
    plt.xlabel("Training Set Size")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.grid()
    plt.show()
```

# Hyperparameter tuning

Random Forest was tuned because it showed the best baseline performance.

GridSearchCV was used to optimize:
- max_depth
- n_estimators
- min_samples_split
- max_features

This improves bias-variance tradeoff systematically.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

param_grid_v2 = {
    'max_depth':         [4, 6, 8],
    'n_estimators':      [100, 200],
    'min_samples_split': [10, 20],
    'max_features':      ['sqrt', 0.5]
}

grid2 = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid_v2, cv=5, scoring='f1', n_jobs=-1
)
grid2.fit(X_train_selected, y_train_res)

print("\nBest Params:", grid2.best_params_)
train_s = grid2.score(X_train_selected, y_train_res)
test_s  = grid2.score(X_test_selected,  y_test)
print(f"Train Score: {train_s:.4f}")
print(f"Test Score:  {test_s:.4f}")
```
